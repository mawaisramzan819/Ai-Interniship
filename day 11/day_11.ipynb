# For loop
### For loop is used to iterate over a sequance such as list, tuple, string or rangfe . It execute a block of code once for each item in the sequance.


```python
n = 4
for i in range(4):
    print(i)
```

# Break
### Break means stop the program immidiatly

```python
numbers = [1,2,3,4,5,6]

for number in numbers:
    if number == 4:
        break 
    print(number)
```

# Continue
### continue means skip the current iteration and move to the next item.

```python
nums = [1,2,3,4,5,6,7,9]

for num in nums:
    if num == 3:
        continue
    print(num)
```

# Enumerate()
### The enumerate() function in python is used to iterate over an iterable while keeping track of index and the value. It rerturns pairs in the form of (index,element). This removes the need to manually maintain a counter variale during iteration.

```python
fruits = ["apple", "banana", "mango", "grapes"]

for fruit in enumerate(fruits, start= 1):
    print(fruit)
```

# Zip()
### The zip() function in python is used to combine two oe morw iterables(like lists, tuples, string, dictionaries,etc) into a single iterator of tuples.

```python
names = ["Awais","Ahmad","Ali  ","Umar "]
marks = [70,75,78,80]

for name,mark in zip(names,marks):
    print(name,mark)
```

# Data processing pipeline

```python
import pprint

students = ["Awais","Ahmad","","Ali","Umar"]
marks = [70,80,90,95,45]
cities = ["Lahore", "Islamabad","Bahawalpur","Faisalabad","Multan"]

PASSING_MARKS = 50
student_dict = {}
if not len(students)  == len(cities) == len(marks):
    print("Error 404! Length of give data is not same.\n")

for index, (student,mark,city) in enumerate(zip(students, marks, cities) , start=1):
    if mark == -999:
        print("Critical error!")
        break
    if student == "":
        print("Invalid student name.\n")
        continue

    print(f"Student {index}: {student} | Marks: {mark}\n City {city}\n") 


    name_upper = student.upper()
    city_upper = city.upper()
    status = "PASS" if mark >= PASSING_MARKS else "FAIL"

    student_dict[f"student {index}"] ={

        "name" : name_upper,
        "city" : city_upper,
        "marks" : mark,
        "status" : status
 
    }
print("\n--- Processed student records dictionary ---\n")

pprint.pprint(student_dict)


```
