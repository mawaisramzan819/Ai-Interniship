# Data types in python 


# What are the data types?
###
"""

Data types are the classification of data items. 
They represent the kind of value that tells what operations can be performed on a particular data.

*Common Types of Data Types in Python*:

### 1.Integer (int): 
      It is used to represent the whole numbers. It can be positive or negative.

      e.g. 1, 2, 3, -1, -2, -3

### 2.Float (float):
        It is used to represent the decimal numbers. It can be positive or negative.


        e.g. 1.5, 2.7, -1.3

### 3.Complex (complex):
         It is used to represent the complex numbers. It has a real and imaginary part. 

        
         e.g. 1+2j, 3-4j
         
### 4.String (str):
         It is used to represent the sequence of characters. It can be enclosed in single, double or triple quotes. 


        e.g. 'Hello', "World", '''Python'''        

"""
###

```python
data1  = 123     # this is integer data type

data2 = "Ali , data , string"       # this is string data type

data3 = True                         # this is boolean data type

data4 = 12.5                           # this is float data type


print("data1 is of type: ", type(data1))
print("data2 is of type: ", type(data2))
print("data3 is of type: ", type(data3))    
print("data4 is of type: ", type(data4))
```

# Practice Questions


```python
first_number = float(input("Enter first number: "))     
second_number = float(input("Enter second number: "))

function = input("Enter the operation you want to perform (+, -, *, /): ")

if function == "+":
    result  = first_number + second_number

    print("The result of addition is:", result)

elif function == "-":
    result  = first_number - second_number

    print("The result of subtraction is:", result)

elif function == "*":
    result  = first_number * second_number

    print("The result of multiplication is:", result)


elif function == "/":
    if second_number == 0:
        print("Error: Division by zero is not allowed.")
    else:
        result  = first_number / second_number
        print("The result of division is:", result)
```
