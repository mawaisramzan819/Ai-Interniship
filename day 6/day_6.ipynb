# List
### List is an ordered, mutable colloections that can hold items of any data

```python
lst = ["apple",3,5,"mango"]
print(lst)
```

# Characteristics of list

## Ordered

### List maintain the order in which items are added

```python
fruits = ["Apple","Banana","Orange"]
print(fruits)
```

## Mutable
### Unlike strings can be modified after creation

```python
fruits = ['Apple','banana','orange']
fruits[-2] = 'mango'
print(fruits)
```

## Allow duplicate

### List can allow duplicate values

```python
numbers = [10,20,20,30,40]
print(numbers)
```

## Types of data

### List can store different type of data

```python
data = ['Awais',20,5.9,True]
print(data)
```

# Methods of list

## Indexing

### Each items has an index

```python
fruits = ['apple','banana','mango']
#           0        1        2 
print(fruits[1])
```

## Negative Indexing

### Pyhton also supports negative indexing

```python
fruits = ['apple','banana','mango']
print(fruits[-1])
```

## List Slicing

### Slicing axtracts part of a list.

```python
data = ['Awais',23,4.4, 'Ali']
print(data[0:2])
```

### python includes the start index but excludes the end index

## append()
### Add one item at the end

```python
name = ['Awais','Ahmad']
name.append('Ali')
print(name)
```

## Insert()
### Add all items from one list

```python
data = ['Awais',2,43,'Ali']
data.insert(3,'Almas')
print(data)
```

## extend()
### Add all items from another list

```python
list1 = [1,2]
list2 = [3,4]
list1.extend(list2)
print(list1)
```

## pop()
### Remove the items by index and returns it


```python
fruits = ['apple','banana','mango']
fruits.pop(1)
print(fruits)
```

## remove()
### This method is used o delete the first occurence of a specified value from a list.

```python
nums = [10,20,30,20]
nums.remove(20)
print(nums)
```

## clear()
### Remove all items from a list. 

```python
numbers = [1,2,3,4,4]
numbers.clear()
print(numbers)
```

## index()
### Index is a Python method and use to find the position of an item in list.

```python
lst = [1,4,6,0,3,23]
print(lst.index(3))
```

## count()
### This method is used to find how many times a value appears in list

```python
data = ['awais',1,3,'awais']
print(data.count('awais'))
```

## sort()
## arrange the item in a list in assending order

```python
num = [2,4,7,3,9,5]
num.sort()
print(num)
```

## reverse()
### Use to reverse the order of items in list

```python
nums = [3,4,5,6]
nums.reverse()
print(nums)
```
