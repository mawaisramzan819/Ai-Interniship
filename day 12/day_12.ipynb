# Function Defination

# (def) keyword
### A function id s reusable block of code tht rund only when you call it.

```python
def greet(name):     # function to  greet the user
    return (f"Hello {name}")   # return the greeting message 
message = greet("Ahmad")       # call the function 
print(message)    # display the greeting message
```

# Perameters:
### Perameters are the variables listed inside the parenthesis of a function defination

```python
def add(a,b):         # a , b are the perameters
    return a +b

func = add (3,4)
print(func)
```

# Arguments:
### When we assign the value to the function while calling the function. The value is known as argument.

```python
def greet(name):
    return (f"Salam {name}")

message = greet("Awais")   # Awais is an argument
print(message)
```

# Positional arguments:
### Positional arguments are value matched to parameter by their position (order) in the function call.

```python
def add (val1,val2):    # positional arguments
    return( val1 + val2)

function = add(34,43)    # positional arguments
print(function)
```

# Keywords arguments:
### Keyword arguments are the values matched to perameters by name not position.

```python
def data(val1,val2):
    return val1 * val2

result = data(val2=74,val1=93)  # Keyword arguments
print(result)
```

# Default perameter:
### They have pre-assigned value in the function defination, so the caller can skip them.

```python
def greet(name= "Awais"):  # Awais is default argument
    return f"Hello {name}"

print(greet())
```

# Local variable:
### A local variable created inside a function. It only exists while the function is running.

```python
def my_function():
    x = 10        # x is Local variable
    x = x +20
    print(x)

my_function()
```

# Global variable:
### A global variable is a variable created outside all function at the module or file level

```python
name = "Awais"   # Global variable

def greet(name):
    print(f"Salam {name}")

greet(name)    
```

# global Keyword:
### The global keyword tells Python that a variable inside a function refers to the global version, allowing you to modify it.

# Note:
### Best practice: Avoid global when possible , instead, pass values as perameters and return results.

```python
x = 0
def modify():
    global x  #  I'm talking about global
    x = x + 1 # Now modifying the global
    print(x)
modify()

# ========================================
#               
# ========================================

def increment(count):
    return count + 1
counter = 0
counter  = increment(counter)
print(f"Initial: {counter}")
```
