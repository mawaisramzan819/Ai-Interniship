```python
def function_name(*args):
    # args is a tuple inside the function
    for item in args:
        print(item)
```

```python
# With *args — caller just passes values naturally
def add(*args):
    total = 0
    for n in args:
        total += n
    return total

add(1, 2, 3)         # natural, no list needed
add(1, 2, 3, 4, 5)   # works with any count
```

```python
def hello(*args, **kwargs):
    print(*args)
    print(**kwargs)

lst = list(("Awais","Ramzan"))
hello(*lst)
dict_val = {('age = 19' , 'dob = 2007')}
hello(**dict_val)

```
