# Sets{}
### A set is an unorderd collection of unique items.

## Key points:
###  Uses curly braces:{1,2,3}
### Does not allow 
### Items are not indexed , so you cannot acess them by position
### Useful for membership testing,removing,duplicates,and methematical set operations

```python
numbers = {1,2,3,2,1,4,5,2}
print(numbers)
print(type(numbers))
```

# Operations of set:


## Add an element : add()

```python
s = {1,2,3}
s.add(4)
print(s)

```

## Remove an element : discard()


```python
s = {1,2,3,4,5}
s.discard(4)
print(s)
```

## Remove all items: clear()

```python
s = {"Awais",1,"Ali"}
s.clear()
print(s)
```

## Union : union()
### This function is used to combine two sets and return a new set with unique values

```python
s1 = {"a","b"}
s2 = {"b","c" }
s = s1.union(s2)
print(s)
```

## Intersection : intersection()
### This function return a new set combining elements that are common

```python
a = {1,2,3,4,5}
b = {2,3,4,6}
i = a.intersection(b)
print(i)
```

## Difference : difference()
### This funtion returns a set containing elements that are in the first set but not in the second.

```python
a = {1,2,3,4,5,6}
b = {2,3,4,5,6,7,8}
d = a.difference(b)
print(d)
```

##  Geographic Coordinates System


```python
import math

lahore = ("Lahore",31.5204,74.3587)
bahawalpur = ("Bahawalpur",29.3957,71.6833)
multan = ("Multan",30.1984, 71.4687)
islamabad = ("Islamabad",33.6844, 73.0479)

cities = (lahore,bahawalpur,multan,islamabad)

def calculate_distance(city1,city2):
    coordinator1 = (city1[1],city1[2])
    coordinator2 = (city2[1],city2[2])
    distance = math.dist(coordinator1,coordinator2)
    return distance

smallest_distance = float("inf")
closest_city = ""

for city in cities:
    if lahore != city:
        distance = calculate_distance(lahore,city)
        if distance < smallest_distance:
            closest_city = city[0]
print("Closest_city is: ",closest_city)
print("Distance is: ", distance)
```
