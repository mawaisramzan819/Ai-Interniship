# Logical operator (and , or , not)


### Logical operator are used for decision making in the program. They can allow to combine and invert multiple conditions to control the flow of the code.

# and Operator

### Output True only if both conditions are true

```python
is_logedin = True  # this variable is used to store the login status of the user
is_balance = 150 
is_withdraw = 100  # this variable is used to store the withdraw status of the user

if is_logedin and is_balance >= is_withdraw:  # this will check if both conditions are true
    print("You can withdraw the money!")  # this will print if both conditions are true
else:
    print("You cannot withdraw the money!")  # this will print if any of the conditions is false
```

# or Operator

### Output True when atleast one condition is true

```python
is_subscription = False
is_voucher = True

if is_subscription or is_voucher:  # this will check if any of the conditions is true
    print("Enjoy your movie!")  # this will print if any of the conditions is true
else:
    print("You cannot watch the movie!")  # this will print if both conditions are false
```

# not Operator

### Flips the value like positive to negative and negative to positive

```python
game_over =False

if not game_over:  # this will check if the condition is false
    print("Game is still on!")  # this will print if the condition is false       
```

# Operator precedence
### When we use multiple operators in single line of code python need set of rules decides which operaton to perform first.

```python
num = 2 + 3 * 5  # this will return 17 because multiplication has higher precedence than addition
print(num)
```

# Simple calculator

```python
def calculator():
    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))
        operator = input("Enter the operator (+,-,*,/): ")

        if operator == "+":
            return num1 + num2
        elif operator == "-":
            return num1 - num2
        elif operator == "*":
            return num1 * num2
        elif operator == "/":
            if num2 == 0:
                return "Cannot divide by zero"
            return num1 / num2
        else:
            return "Invalid operator"
    except ValueError:
        return "Invalid input. Please enter numeric values."

print(calculator())
```
