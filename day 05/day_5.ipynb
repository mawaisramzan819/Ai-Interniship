# Reinforcement learning
### The model learn through trial and errors. Good action = reward , Bad action = panalties
### e.g (Self driving car , games feedeback)

```python
reward = 10
if reward > 0:
    print("Good move")
else:
    print("Bad move")    
```

# Classification
### It predicts catagories of data.
### e.g (spam or not spam , adult or minor)


```python
age  = 20

if age > 18:
    print("Adult")
else:
    print("Minor")    
```

# Decision Tree
### A decision tree is a flowchart - like algorithm. It asks question on each step until final step reached

```python
age = 20
license = True

if age >= 18:
    if license:
        print("Can drive")
    else:
        print("License Required!")
else:
    print("Too young")   
```
