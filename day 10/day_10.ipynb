# Dictionaries { }
### A dictionary is a collection of data that store in the form of *key* - *Value* pairs where each key act as a unique lable to identify and retrive its value.

### Think of a python is like a contact in your phone. In a contact list you look up a person name and find his phone number.In python name is called a key and number is called a value. Together they form a key value pair.

```python
contact = {
    "Alice" : 12345,
    "john" : 67890
}
print(contact)
```

## Adding a new key

```python
data = {
    "Awais" : 11223344,
    "Umar" : 44332211,
}
# Adding a new key value pair

data["Ali"] = 55667788

print(data)
print(data["Ali"])
```

## Sorted () Function:
### The sorted() function return a new list containing a dictionary,s keys in a sorted order.
## Important:
### By default , it sorted the key the value. In simple way sorted means or order the value accuratly. Keys are arranged alphabetically.

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Python"
}

print(data)
print(sorted(data))
```

# Max ( ):
### Max functions return the largest key in the dictionary according to python comperison rules (alphabetically for  strings)

```python
students = {
    "Awais" : 70,
    "Ahmad" : 85,
    "Bilal" : 90
}

print(max(students))                 # It returns bilal max funtion check the keys not values, in (A,B) B come after A then it will return Bilal as max key
print(max(students.values()))    # Using .values() function it return max value in the keys
print(students)
```

# min( ):
### min( ) function return the smallest key in the dictionary

```python
students = {
    "Awais" : 70,
    "Ahmad" : 85,
    "Abbas" : 90
}

print(min(students))
print(min(students.values()))
```

# Sum( ):
### The sum function return sum of all numeric keys in the dictionary

```python
num = {
    1:"one",
    2:"two",
    3: "three",
    4: "Four"
}
print(sum(num))
```

# Dictionary Methods:
### Dictionary methods are the built in funtions that belongs to dictionary (dict) class and used to perform actions such as accessing, adding, updating,copying and removing data from a dictionary.

## get( ):
### The get() method is return the value of the specific key.If the key is not exist in the dictionary it return *None* instead of raising an error.

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Ai engineering"
}
print(data.get("name"))
print(data.get("city"))
```

## keys( ) : 
### The keys () function return the keys in which the values are given.

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Ai engineering"
}
print(data.keys())
```

## items ( ):
### The items() method return a view object containing all the key - value as a tuple

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Ai engineering"
}
print(data.items())
```

## values( ) : 
### The values() return the values of the keys.

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Ai engineering"
}
print(data.values())
```

## update( ):
### The update() method add new key - value pairs and update the existing keys

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Ai engineering"
}
data.update({"work" : "Xeven"})
data.update({"age" : 20})
print(data)
```

## pop( ):
###    The pop() method is remove the specific key and return its value.

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Ai engineering"
}
print(data.pop("age"))
```

## popitem( ):
### The popitem() remove and return the last inserted key value pair

```python
data = {
    "name" : "Awais",
    "age" : 19,
    "course" : "Ai engineering"
}

data.popitem()
print(data)
```

# Nested Dictionary:
### A dictionary inside the aonther dictionary is known as nested dictionary. It help us to organize related information together.

```python
data = {
    "Ali" : {
        "age" : 18,
        "course"  : "python",
        "city"  : "Lahore"
    } ,
    "Ahmad"  : {
        "age" : 19,
        "course" : "Ai engineering",
        "city"  : "Bahawalpur"
    }
}
print(data)                  # ouput all the dictionary
print(data["Ahmad"])           #  output Ahmad data store in another dictionary
print(data["Ali"]["age"])      # ouput that show just age of Ali which is given in nested dictionary 
```

```python
data = {
    "Alice" : {
        "age" : 19,
        "grade" : "A+",
        "city"  : "New York"
    
    },
    "John" : {
        "age" : 20,
        "grade"  : "B",
        "city" : "London"
    }
}
del data ["Alice"]["age"]  # Remove one detail
print(data["Alice"])
del data["Alice"]  # Remove entire entity
print(data)
```

# Student Information System

```python
import json
from os import name 

student = {
    "student_id1" : {
        "name"  : "Awais",
        "age"  : 19,
        "grades"  : {
            "Physics" : "A",
            "Maths" : "B",
            "English"  : "A"

        }
    }
}
student["student_id2"] = {
    "name"  :"Ahmad" ,
    "age"  : 18,
    "grades"  : {
        "Physics"  : "B",
        "Maths"  : "B",
        "English"  : "A+"
    }
}
student["student_id1"]["grades"] = {
    "Physics" : "B",
    "Maths" : "A",
    "English" : "A+"
}
grade_points = {"A+":4, "A": 3.5 , "B" : 3 , "C" : 2.5}

for sid,info in student.items():
    grades = info["grades"]

    print("StudentID: ", sid)
    print("=" * 60)
    print("Name: ",info["name"])
    print("Age: ",info["age"])
    print("Grade: ",info["grades"])

    average = sum((grade_points[g] for g in grades.values())) / len(grades)
    print(f"The GPA of {sid } {info['name']} is:" , round(average , 2))
    print("=" * 60)

top_student = max(student , key=lambda sid: average) 
print(f"Top student is: {student[top_student]['name']}")
print("=" * 60)   

with open("student.json","w") as file:
    json.dump(student,file,indent=4)

with open("student.json","r") as file:
   data = json.load(file)
   print(data) 
```
