# Conditional statment


### Conditional statment






A piece of code that runs different actions depending on wether a condition is true or false

For example
"If this is true do X, Otherwise do Y"

```python
age = int(input("Enter your age: "))  # age is a variable that will store the user input as an integer

if age >= 18:                          # this is conditional statment 
    print("You are eligible to vote.") 
else:
    print("You are not eligible to vote.") # this is the else block
```

# Boolean Logic

It is a system of resoning that work with only two values (*True and False*)

```python
age  = int(input("Enter your age:"))
has_license = input("Do you have a driving license? (True/False):") == "True" .strip().lower()

if age >=18 and  has_license:    # and operator is used to check if both conditions are true
    print("You can drive a car")
else:
    print("You cannot drive a car")    
```

# Double equal sign (==)

### It is used for check the values are equal or not.

```python
num1 = 12
num2 = 12
num1 == num2  # this will return True because both numbers are equal

val1 = "Str"
val2 = "int"

val1 == val2  # this will return False because both values are not equal
```

# Nested if Statment

### An if statment tucked inside the another if statment.


```python
is_member = True
is_birthday = True

if is_member == True:
    print("Welcome, member!")

    if is_birthday == False:  # This is nested if statement that checks if the user is a member and if it's their birthday
        print("Happy Birthday! You get a special discount.") 
    else:
        print("Enjoy your shopping!")

else:
    print("Welcome, member! Signup today for saving.")
```

# Age verification

```python
try:
    age = int(input("Enter your age: "))

    if age < 0:
        print("age cannot be negative.")  # This will raise an error if the age is negative
    elif age < 18:
        print("You are minor.")
    elif age >= 18 and age < 65:
        print("You are an adult.")
    elif age >= 65:
        print("You are a senior citizen.")
    elif age >= 65:
        print("You are an old citizen.") 
    else:
        print("Invalid age input.")

except ValueError:
    print("Invalid input. Please enter a valid age.")  # This will handle the error if the user enters a non-integer value   

```

# Eligible for Driving
